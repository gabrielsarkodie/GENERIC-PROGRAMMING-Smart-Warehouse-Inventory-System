```python
from typing import Generic, TypeVar

# Define a TypeVariable to allow any data type
T = TypeVar("T")


class Storage(Generic[T]):

    def __init__(self):
        self._item: T | None = None

    def store(self, item: T) -> None:
        """Stores one item of type T."""
        self._item = item

    def retrieve(self) -> T | None:
        """Retrieves the stored item."""
        return self._item


# --- Demonstrating the Class ---

# 1. Using an Integer
int_storage = Storage[int]()
int_storage.store(42)
print("Stored Integer:", int_storage.retrieve())

# 2. Using a String
string_storage = Storage[str]()
string_storage.store("Smart Warehouse Package #A109")
print("Stored String:", string_storage.retrieve())

# 3. Using a List
list_storage = Storage[list[str]]()
list_storage.store(["Item A", "Item B", "Item C"])
print("Stored List:", list_storage.retrieve())
```
